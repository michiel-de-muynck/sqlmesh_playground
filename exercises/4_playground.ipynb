```python
import matplotlib.pyplot as plt
import notebook_utils

df = notebook_utils.query_duckdb(
    """
    SELECT * FROM persistent.gharchive__dev.commits_stats_per_hour
    ORDER BY hour
    """
)

plt.figure(figsize=(10, 6))
plt.plot(df['hour'], df['num_commits'], marker='o', linestyle='-')
plt.title('Number of Commits Over Time')
plt.xlabel('Time')
plt.ylabel('Number of Commits')
plt.grid(True)
plt.xticks(rotation=45)
plt.show()
```
